```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
import os
current_path = os.path.dirname(os.path.abspath("."))
if not os.path.exists('output'):
    os.makedirs('output')
OUTPUT_PATH = os.path.abspath('output')
```

```python
def get_execution_time(appname, checkpoint=False, debloat=True):
    BASE_PATH = f'{appname}'
    filename = "criu_" if checkpoint else ""
    filename += "debloated" if debloat else "normal"
    try:
        with open(BASE_PATH + f"/{filename}.txt") as f:
            lines = f.readlines()
            return np.mean([float(x) for x in lines])
    except:
        return 0
```

```python
def gather_all(appname):
    normal = get_execution_time(appname, False, False)
    debloated = get_execution_time(appname, False, True)
    criu_normal = get_execution_time(appname, True, False)
    criu_debloated = get_execution_time(appname, True, True)
    return normal, debloated, criu_normal, criu_debloated
```

```python
def plot_bar(application):
    
    normal, debloated, criu_normal, criu_debloated = gather_all(application)

    fig, ax = plt.subplots(figsize=(10, 6))
    width = 0.1
    ind = np.arange(1)
    bars1 = ax.bar(ind, [normal], width, label='Original', edgecolor='black', color='C0')
    bars2 = ax.bar(ind + width, [debloated], width, label='λ-TRIM', edgecolor='black', color='C1', hatch='..')
    bars3 = ax.bar(ind + 2*width, [criu_normal], width, label='criu', edgecolor='black', color='C2', hatch='OO')
    bars4 = ax.bar(ind + 3*width, [criu_debloated], width, label='criu + λ-TRIM', edgecolor='black', color='C3', hatch='**')

    ax.set_xticks(ind + 1.5*width)
    ax.set_xticklabels([application])
    ax.legend()
    plt.ylabel("Import time (s)")
    plt.title(application)
    plt.grid(True, linestyle='--', alpha=0.2)

    # Adding value labels on top of the bars
    for bars in [bars1, bars2, bars3, bars4]:
        for bar in bars:
            yval = bar.get_height()
            ax.text(bar.get_x() + bar.get_width() / 2, yval, f'{yval:.3f}', ha='center', va='bottom')

    plt.savefig(OUTPUT_PATH + f'/{application}.png')
    plt.show()
```

```python
# given a list of applications, plot a common bar chart, with 4 bars for each application
def plot(applications, figsize=None):
    applications.sort()
    if figsize is None:
        fig, ax = plt.subplots()
    else:
        fig, ax = plt.subplots(figsize=figsize)
    width = 0.2
    ind = np.arange(len(applications))
    # pattern
    bars1 = ax.bar(ind, [get_execution_time(app, False, False) for app in applications], width, label='Original', edgecolor='black', color='C0')
    bars2 = ax.bar(ind + width, [get_execution_time(app, False, True) for app in applications], width, label='λ-TRIM', edgecolor='black', color='C1', hatch='..')
    bars3 = ax.bar(ind + 2*width, [get_execution_time(app, True, False) for app in applications], width, label='criu', edgecolor='black', color='C2', hatch='OO')
    bars4 = ax.bar(ind + 3*width, [get_execution_time(app, True, True) for app in applications], width, label='criu + λ-TRIM', edgecolor='black', color='C3', hatch='**')

    ax.set_xticks(ind + 1.5*width)
    ax.set_xticklabels(applications, rotation=45, ha='right')
    # make legend a little bigger
    
    ax.legend(fontsize=12)
    plt.ylabel("Import time (s)")
    plt.title("Import time for different applications")
    plt.grid(True, linestyle='--', alpha=0.2)

    plt.show()
```

```python
plot_bar("igraph")
```

```python
applications = [
    "dna-visualization", 
    "tensorflow", 
    "skimage", 
    "lxml",
    "epub-pdf",
    "scikit",
    "pandas",
    "spacy", 
    "lightgbm", 
    "markdown", 
    "chdb-olap", 
    "jsym",
    "textblob",
    "igraph",
    "resnet",
    "wine",
    "image-resize",
    "shapely-numpy", 
    "huggingface", 
    "qiskit-nature"
]
applications.sort()
plot(applications, (20, 6))
```
